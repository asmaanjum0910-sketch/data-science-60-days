```python
import pandas as pd

df = pd.read_csv('IndianHouses.csv')

x = df['Bathroom'].mean()#Mean = the average value (the sum of all values divided by number of values).

df['Bathroom'].fillna(x, inplace=True) #The fillna() method allows us to replace empty cells with a value:

print(df.to_string())
```
