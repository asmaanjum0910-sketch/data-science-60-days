Plot a graph between BHK of Apartment and Builder Floor

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('IndianHouses.csv')

builder_floor = df[df['Type'] == 'Builder_Floor']

x = builder_floor['BHK'].value_counts().sort_index().index
y = builder_floor['BHK'].value_counts().sort_index().values

plt.xlabel('BHK')
plt.ylabel('Number of Builder Floor Houses')
plt.title('BHK vs Builder Floor')

plt.plot(x, y, '--')

plt.grid(axis='y', linestyle='-.', linewidth=1)

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('IndianHouses.csv')

apartment = df[df['Type'] == 'Apartment']
builder_floor = df[df['Type'] == 'Builder_Floor']

x = sorted(df['BHK'].dropna().unique())

y_apartment = apartment['BHK'].value_counts().reindex(x, fill_value=0)
y_builder = builder_floor['BHK'].value_counts().reindex(x, fill_value=0)

plt.plot(x, y_apartment, '--', label='Apartment')
plt.plot(x, y_builder, '--', label='Builder Floor')

plt.xlabel('BHK')
plt.ylabel('Number of Houses')
plt.title('BHK of Apartment and Builder Floor')

plt.grid(axis='y', linestyle='-.', linewidth=1)
plt.legend()

plt.show()
```

subplot

1.   subplot() function you can draw multiple plots in one figure:
2.   plt.subplot(1, 2, 1)
#the figure has 1 row, 2 columns, and this plot is the first plot.

plt.subplot(1, 2, 2)
#the figure has 1 row, 2 columns, and this plot is the second plot.



Form a total of 6 plots (2 rows) in one figure.

The first plot should be Bathrooms of Apartment vs BHK of Apartment

The second will be Status of new property and status of Resale

The third one per sqft of Furnished vs Semifurnished

The fourth - Price vs per sqft

The fifth - Ready to move Areas vs prices

The sixth - Parking vs BHK

```python
#plot_1
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
df=pd.read_csv('IndianHouses.csv')
x=df['Bathroom']
y=df['BHK']
plt.subplot(2,3,1)
plt.xlabel('Bathroom')
plt.ylabel('BHK')
plt.title('Bathroom vs BHK')
plt.plot(x,y,'-.')

#plot_2--The second will be Status of new property and status of Resale
Status_new_property = df[df['Transaction'] == 'New_Property']
Status_Resale = df[df['Transaction'] == 'Resale']

x = ['New_Property', 'Resale']

y = [
    len(Status_new_property),
    len(Status_Resale)
]

plt.subplot(2,3,2)

plt.xlabel('Transaction')
plt.ylabel('Number of Houses')
plt.title('New Property vs Resale')

plt.plot(x, y, '--')

#plot_3--The third one per sqft of Furnished vs Semifurnished
furnished = df[df['Furnishing'] == 'Furnished']
semi_furnished = df[df['Furnishing'] == 'Semi-Furnished']

x1 = furnished['BHK']
y1 = furnished['Per_Sqft']

x2 = semi_furnished['BHK']
y2 = semi_furnished['Per_Sqft']

plt.subplot(2,3,3)

plt.plot(x1, y1, '--', label='Furnished')
plt.plot(x2, y2, '--', label='Semi-Furnished')

plt.xlabel('BHK')
plt.ylabel('Per Sqft')
plt.title('Per Sqft: Furnished vs Semi-Furnished')
plt.legend() #- adds a legend/key to your graph. It tells you which line or marker represents which data.

#plot_4-The fourth - Price vs per sqft
x=df['Price']
y=df['Per_Sqft']
plt.subplot(2,3,4)
plt.xlabel('Price')
plt.ylabel('Per_Sqft')
plt.title('Price vs Per_Sqft')
plt.plot(x,y,'o')

#plot_5-The fifth - Ready to move Areas vs prices
ready_to_move_df = df[df['Status'] == 'Ready_to_move']

x = ready_to_move_df['Area']
y = ready_to_move_df['Price']

plt.subplot(2,3,5)

plt.xlabel('Area')
plt.ylabel('Price')
plt.title('Ready to Move Areas vs Prices')

plt.plot(x, y, '--')

#plot_6- The sixth - Parking vs BHK
x=df['Parking']
y=df['BHK']
plt.subplot(2,3,6)
plt.xlabel('Parking')
plt.ylabel('BHK')
plt.title('Parking vs BHK')
plt.plot(x,y,'o')
```

Display the Area vs Price of top 10 expensive houses in different sizes. The most expensive should have the biggest size.

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('IndianHouses.csv')

top10 = df.sort_values('Price', ascending=False).head(10)

x = top10['Area']
y = top10['Price']

size = top10['Price'] / top10['Price'].max() * 500

plt.scatter(x, y, s=size, color='red')

plt.xlabel('Area')
plt.ylabel('Price')
plt.title('Area vs Price of Top 10 Expensive Houses')

plt.show()
```
