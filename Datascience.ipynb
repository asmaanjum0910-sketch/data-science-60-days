subplot

2.   plt.subplot(1, 2, 1)
#the figure has 1 row, 2 columns, and this plot is the first plot.

plt.subplot(1, 2, 2)

Form a total of 6 plots (2 rows) in one figure.

The first plot should be Bathrooms of Apartment vs BHK of Apartment

The second will be Status of new property and status of Resale

The third one per sqft of Furnished vs Semifurnished

The fourth - Price vs per sqft

The fifth - Ready to move Areas vs prices

The sixth - Parking vs BHK

```python
#plot_1
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
df=pd.read_csv('IndianHouses.csv')
x=df['Bathroom']
y=df['BHK']
plt.subplot(2,3,1)
plt.xlabel('Bathroom')
plt.ylabel('BHK')
plt.title('Bathroom vs BHK')
plt.plot(x,y,'-.')

#plot_2--The second will be Status of new property and status of Resale
Status_new_property = df[df['Transaction'] == 'New_Property']
Status_Resale = df[df['Transaction'] == 'Resale']

x = ['New_Property', 'Resale']

y = [
    len(Status_new_property),
    len(Status_Resale)
]

plt.subplot(2,3,2)

plt.xlabel('Transaction')
plt.ylabel('Number of Houses')
plt.title('New Property vs Resale')

plt.plot(x, y, '--')

#plot_3--The third one per sqft of Furnished vs Semifurnished
furnished = df[df['Furnishing'] == 'Furnished']
semi_furnished = df[df['Furnishing'] == 'Semi-Furnished']

x1 = furnished['BHK']
y1 = furnished['Per_Sqft']

x2 = semi_furnished['BHK']
y2 = semi_furnished['Per_Sqft']

plt.subplot(2,3,3)

plt.plot(x1, y1, '--', label='Furnished')
plt.plot(x2, y2, '--', label='Semi-Furnished')

plt.xlabel('BHK')
plt.ylabel('Per Sqft')
plt.title('Per Sqft: Furnished vs Semi-Furnished')
plt.legend() #- adds a legend/key to your graph. It tells you which line or marker represents which data.

#plot_4-The fourth - Price vs per sqft
x=df['Price']
y=df['Per_Sqft']
plt.subplot(2,3,4)
plt.xlabel('Price')
plt.ylabel('Per_Sqft')
plt.title('Price vs Per_Sqft')
plt.plot(x,y,'o')

#plot_5-The fifth - Ready to move Areas vs prices
ready_to_move_df = df[df['Status'] == 'Ready_to_move']

x = ready_to_move_df['Area']
y = ready_to_move_df['Price']

plt.subplot(2,3,5)

plt.xlabel('Area')
plt.ylabel('Price')
plt.title('Ready to Move Areas vs Prices')

plt.plot(x, y, '--')

#plot_6- The sixth - Parking vs BHK
x=df['Parking']
y=df['BHK']
plt.subplot(2,3,6)
plt.xlabel('Parking')
plt.ylabel('BHK')
plt.title('Parking vs BHK')
plt.plot(x,y,'o')
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('IndianHouses.csv')

types = df['Type'].unique()
sizes = [30, 80, 150, 250, 350]
markers = ['o', 's', '^', 'D', '*']

for i, house_type in enumerate(types):
    data = df[df['Type'] == house_type]

    plt.scatter(
        data['Area'],
        data['Price'],
        marker=markers[i],
        alpha=0.5,
        s=sizes[i],
        label=house_type
    )

plt.xlabel('Area')
plt.ylabel('Price')
plt.title('Area vs Price by House Type')

plt.legend(title='House Type')

plt.show()
```

```python
import pandas as pd
from word2number import w2n #!pip install word2number - first run this and import

# =========================================================
# Read CSV
# =========================================================

df = pd.read_csv('Diabetes-Dataset (1).csv')


# =========================================================
# Patient_ID
# =========================================================

# Replace empty Patient_ID with 151
df.fillna({"Patient_ID": 151}, inplace=True)

# Convert Patient_ID to integer
df['Patient_ID'] = df['Patient_ID'].astype(int)


# =========================================================
# Age
# =========================================================

# Replace word "Thirty" with 30
df['Age'] = df['Age'].replace('Thirty', 30)

# Convert Age to numeric
df['Age'] = pd.to_numeric(df['Age'], errors='coerce')

# Replace unrealistic Age values
df['Age'] = df['Age'].replace([-5, 150], 30)

# Remove empty Age cells
df.dropna(subset=['Age'], inplace=True)


# =========================================================
# BMI
# =========================================================

# Convert BMI to numeric
df['BMI'] = pd.to_numeric(df['BMI'], errors='coerce')

# Remove rows where BMI is empty
df.dropna(subset=['BMI'], inplace=True)


# =========================================================
# Gender
# =========================================================

# Remove extra spaces and convert everything to lowercase
df['Gender'] = df['Gender'].astype(str).str.strip().str.lower()

# Fix incorrect Gender values,['nan': pd.NA,]-Converts the text "nan" into a real missing value,['': pd.NA]-Converts an empty string into a missing value.
df['Gender'] = df['Gender'].replace({
    'mle': 'male',
    'male': 'male',
    'female': 'female',
    'femle': 'female',
    'other': 'other',
    'nan': pd.NA,
    '': pd.NA
})


# =========================================================
# Average BMI
# =========================================================

avg_bmi = round(df['BMI'].mean())

print("Average BMI:", avg_bmi)


# =========================================================
# Replace unrealistic BMI values
# =========================================================

for index, row in df.iterrows():

    if row['BMI'] < 10 or row['BMI'] > 60:

        df.loc[index, 'BMI'] = avg_bmi


# Replace empty BMI values
df['BMI'] = df['BMI'].fillna(avg_bmi)


# =========================================================
# Convert Other / Blank Gender using BMI
# =========================================================

for index, row in df.iterrows():

    if pd.isna(row['Gender']):
        gender = True
    elif row['Gender'] == 'other':
        gender = True
    else:
        gender = False

    if gender:

        if row['BMI'] <= 29.5:
            df.loc[index, 'Gender'] = 'male'
        else:
            df.loc[index, 'Gender'] = 'female'


# Convert Male/Female to proper capitalization
df['Gender'] = df['Gender'].replace({
    'male': 'Male',
    'female': 'Female'
})


# Check Gender values
print("\nGender values:")
print(df['Gender'].value_counts(dropna=False))

print("\nUnique Gender values:")
print(df['Gender'].unique())


# =========================================================
# Glucose
# =========================================================

for index, value in df['Glucose_mg_dL'].items():

    # Skip empty values
    if pd.isna(value):
        continue
    #Converts the value to text and removes extra spaces.
    value = str(value).strip()

    # If value is already a number
    if value.isdigit():

        df.loc[index, 'Glucose_mg_dL'] = int(value)

    else:

        # Convert words such as "one hundred" to number
        try:

            df.loc[index, 'Glucose_mg_dL'] = w2n.word_to_num(value)

        except:

            continue


# Convert Glucose column to numeric
df['Glucose_mg_dL'] = pd.to_numeric(
    df['Glucose_mg_dL'],
    errors='coerce'
)


# =========================================================
# Average Glucose
# =========================================================

Glucose_avg = int(round(df['Glucose_mg_dL'].mean()))

print("\nAverage Glucose:", Glucose_avg)


# Replace unrealistic value
df['Glucose_mg_dL'] = df['Glucose_mg_dL'].replace(
    [-15],
    Glucose_avg
)


# Replace empty Glucose values
df['Glucose_mg_dL'] = df['Glucose_mg_dL'].fillna(Glucose_avg)


# Convert Glucose to float
df['Glucose_mg_dL'] = df['Glucose_mg_dL'].astype(float)


# =========================================================
# HbA1c
# =========================================================

# Convert HbA1c to numeric
df['HbA1c'] = pd.to_numeric(
    df['HbA1c'],
    errors='coerce'
)


# =========================================================
# Average HbA1c
# =========================================================

avg_HbA1c = round(df['HbA1c'].mean(), 2)

print("Average HbA1c:", avg_HbA1c)


# =========================================================
# Replace unrealistic HbA1c
# =========================================================

for index, row in df.iterrows():

    if row['HbA1c'] < 3 or row['HbA1c'] > 10:

        df.loc[index, 'HbA1c'] = avg_HbA1c


# Replace empty HbA1c
df['HbA1c'] = df['HbA1c'].fillna(avg_HbA1c)


# =========================================================
# Final Data
# =========================================================

print("\nCleaned Data:")
print(df.to_string())


# =========================================================
# Check Missing Values
# =========================================================

print("\nMissing values:")
print(df.isna().sum())


# =========================================================
# Save Cleaned DataFrame
# =========================================================

df.to_csv('Diabetes-Cleaned.csv', index=False)


# =========================================================
 #Download CSV in Google Colab
# =========================================================

from google.colab import files

files.download('Diabetes-Cleaned.csv')
```
