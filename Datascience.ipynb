correclation - gives metrics values
how two numerical columns are related to each other
| Correlation | Meaning                       |
| ----------: | ----------------------------- |
|      **+1** | Perfect positive relationship |
|    **+0.8** | Strong positive               |
|    **+0.5** | Moderate positive             |
|       **0** | No linear relationship        |
|    **-0.5** | Moderate negative             |
|    **-0.8** | Strong negative               |
|      **-1** | Perfect negative              |

fetch all locality names that start with 'k'

```python
df=df[df['Locality'].str.contains('^k', case=False, na=False)]
print(df['Locality'])
```

Get me the number of Ready to Move Apartments

```python
ready_to_move_df = df[df['Status'] == 'Ready_to_move']
number_of_ready_to_move_apartments = len(ready_to_move_df)
print(number_of_ready_to_move_apartments)
```

```python
number_of_ready_to_move_apartments = (df['Status'] == 'Ready_to_move').sum()
print(number_of_ready_to_move_apartments)
```

```python
import matplotlib.pyplot as plt
import pandas as pd
df=pd.read_csv('IndianHouses.csv')
x=df['BHK']
y=df['Bathroom']
plt.plot(x,y,'o')
plt.xlabel('BHK')
plt.ylabel('Bathroom')
plt.title('BHK vs Bathroom')
plt.show()
```
